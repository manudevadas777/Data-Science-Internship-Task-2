# Titanic Survival Analysis
## Data Science with Python Internship - Task 2

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Titanic-Dataset.csv")
df.head()
```

```python
df.info()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df.isnull().sum()
```

```python
df = df.drop(columns=["Cabin"])
```

```python
df.isnull().sum()
```

```python
survival_gender = df.groupby("Sex")["Survived"].mean()
print(survival_gender)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))
sns.barplot(x="Sex", y="Survived", data=df)

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")

plt.show()
```

```python
survival_class = df.groupby("Pclass")["Survived"].mean()
print(survival_class)
```

```python
plt.figure(figsize=(6,4))
sns.barplot(x="Pclass", y="Survived", data=df)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")

plt.show()
```

```python
bins = [0, 12, 18, 35, 60, 100]
labels = ["Child", "Teen", "Young Adult", "Adult", "Senior"]

df["AgeGroup"] = pd.cut(df["Age"], bins=bins, labels=labels)
```

```python
age_survival = df.groupby("AgeGroup")["Survived"].mean()
print(age_survival)
```

```python
plt.figure(figsize=(8,4))
sns.barplot(x="AgeGroup", y="Survived", data=df)

plt.title("Survival Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Survival Rate")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Age"], bins=20, kde=True)

plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Count")

plt.show()
```

## Conclusion

- Female passengers had a significantly higher survival rate than male passengers.
- First-class passengers had the highest survival rate, while third-class passengers had the lowest.
- Children had the highest survival rate among the age groups, whereas senior passengers had the lowest.
- The passenger age distribution shows that most travelers were between 20 and 40 years old.
